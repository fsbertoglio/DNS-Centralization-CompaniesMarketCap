### ✅ P5. As empresas brasileiras mantêm algum grau de controle direto sobre sua infraestrutura de DNS (NS próprio, ASN próprio ou operador nacional)? (entrar na 6)

#### 🎯 Objetivo da Análise
Investigar se as empresas operam seus próprios servidores DNS ou se dependem exclusivamente de terceiros, e em que medida esse controle está associado à presença de ASN próprio ou infraestrutura nacional.

#### 📊 Estrutura dos Dados Necessária

**Dados por domínio (consolidados):**

| domain           | owns_asn | all_ns_in_own_asn | uses_national_asn | only_third_party_ns | inferred_control_level    |
|------------------|----------|-------------------|-------------------|---------------------|--------------------------|
| empresa1.com.br  | True     | True              | True              | False               | Total                    |
| empresa2.com.br  | False    | False             | False             | True                | Nenhum                   |
| empresa3.com.br  | False    | False             | True              | False               | Parcial (nacional)       |

- **owns_asn**: True se algum NS tem ASN cuja descrição contém o nome da empresa.
- **uses_national_asn**: True se algum ASN for brasileiro (ex: país = BR).
- **only_third_party_ns**: True se todos os NSs pertencem a provedores como Google, Cloudflare, Amazon, etc.

#### 🛠️ Como calcular os indicadores

A partir do DataFrame de NSs por domínio:

1. **Verificar se o ASN corresponde à empresa:**  
    Ter uma lista com o nome da empresa e verificar se ele aparece parcial ou totalmente em `org_name` ou `as_name`.

2. **Avaliar se há ASNs nacionais e presença de NSs próprios:**  
    - `owns_asn`: True se algum NS tem ASN cuja descrição contém o nome da empresa.
    - `uses_national_asn`: True se algum ASN for brasileiro (ex: país = BR).
    - `only_third_party_ns`: True se todos os NSs pertencem a provedores como Google, Cloudflare, Amazon, etc.

#### 📈 Visualizações recomendadas para P5

1. **Gráfico de pizza:** Nível de controle da infraestrutura de DNS  
    *Categorias: Total / Parcial (ASN nacional ou próprio) / Nenhum*

2. **Gráfico de barras por macro setor:**  
    *Avaliar que setores concentram mais ou menos autonomia técnica*

3. **Tabela:** Top 20 empresas que operam sua própria infraestrutura

#### 🧠 Interpretação esperada

Empresas com controle direto (own ASN) possuem maior independência técnica e soberania digital.

Aquelas com apenas terceirização estão mais sujeitas a:
- Falhas externas
- Bloqueios jurídicos
- Restrições de uso de infraestrutura estrangeira

Essa análise serve como insumo para discussões sobre estratégias de nacionalização da infraestrutura crítica e melhores práticas de governança tecnológica.

#### 🧩 Complementos possíveis

- Associar com localização física do ASN e do datacenter, se disponível.
