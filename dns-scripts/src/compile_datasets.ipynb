```python
import pandas as pd

BASE_DATA = r"..\src\br_merged_data.csv"

base_data_df = pd.read_csv(BASE_DATA, sep=';', encoding='utf-8')

def classify_infra(countries):
    countries = set([c for c in countries if pd.notnull(c)])
    if not countries:
        return 'desconhecido'
    if countries == {'BR'}:
        return '100% nacional'
    elif 'BR' in countries:
        return 'mista'
    else:
        return '100% estrangeira'

# Agrupar e agregar
grouped = base_data_df.groupby('company_name').agg(
    country_name=('country_name', 'first') , # País da empresa minusculo
    state_name=('state_name', 'first'), # Estado da empresa
    city_name=('city_name', 'first'), # Cidade da empresa
    sector_primary=('sector_primary', 'first'), # Setor primário da empresa
    revenue=('revenue', 'first'), # Receita anual em milhões de reais
    earnings=('revenue', 'first'), # Receita anual em milhões de reais
    total_assets=('total_assets', 'first'), # Ativos totais em milhões de reais
    domain=('domain', 'first'), # Domínio analisado
    num_ns=('ns', 'nunique'), # Quantidade total de NSs
    num_asn=('asn', 'nunique'), # Quantidade de ASNs distintos
    asn_list=('asn', lambda x: sorted(set(str(a) for a in x if pd.notnull(a)))), # Lista de ASNs distintos
    num_country=('as_country', 'nunique'), # Quantidade de países distintos
    countries_used=('as_country', lambda x: sorted(set(c for c in x if pd.notnull(c)))), # Lista de países distintos
    pct_national_ns=('as_country', lambda x: sum(c == 'BR' for c in x if pd.notnull(c)) / len(x) if len(x) > 0 else 0), # Percentual de NSs no Brasil
    only_foreign_ns=('as_country', lambda x: all(c != 'BR' for c in x if pd.notnull(c))), # Se todos os NSs são estrangeiros
    has_ipv4_ns=('ns_a', lambda x: any(x)), # Se há NSs com IPv4
    has_ipv6_ns=('ns_aaaa', lambda x: any(x)), # Se há NSs com IPv6
    dual_stack=('ns_a', lambda x: any(x) and any(base_data_df.loc[x.index, 'ns_aaaa'])), # Se há suporte a IPv4 e IPv6
    dominant_provider=('org_name', lambda x: x.mode()[0] if not x.mode().empty else None), # Provedor dominante
    dominant_provider_country=('as_country', lambda x: x.mode()[0] if not x.mode().empty else None), # País do provedor dominante
    dominant_provider_asn=('asn', lambda x: x.mode()[0] if not x.mode().empty else None), # ASN do provedor dominante
    only_one_provider=('org_name', lambda x: len(set(x)) == 1), # Se todos os NSs são do mesmo provedor
    distinct_asns=('asn', lambda x: len(set(a for a in x if pd.notnull(a)))), # Quantidade de ASNs distintos
    distinct_countries=('as_country', lambda x: len(set(c for c in x if pd.notnull(c)))), # Quantidade de paí ses distintos
    country_list=('as_country', lambda x: sorted(set(c for c in x if pd.notnull(c)))), # Lista de países distintos
    provider_list=('org_name', lambda x: sorted(set(o for o in x if pd.notnull(o)))), # Lista de provedores distintos
    num_ns_in_brazil=('as_country', lambda x: sum(c == 'BR' for c in x if pd.notnull(c))), # Quantidade de NSs no Brasil
    num_ns_foreign=('as_country', lambda x: sum(c != 'BR' and pd.notnull(c) for c in x)) # Quantidade de NSs estrangeiros
).reset_index()

# Inferir tipo de infraestrutura
grouped['infra_type'] = grouped['country_list'].apply(classify_infra) # Classificação da infraestrutura

grouped.to_csv(r'.\br_analysis_data.csv', sep=';', index=False, encoding='utf-8')

display(grouped.head())
```
