```python
import dns.resolver
from ipwhois import IPWhois

domain = "petrobras.com.br"

def safe_query(domain, record_type=""):
    try:
        if record_type:
            return [r.to_text() for r in dns.resolver.resolve(domain, record_type)]
        else:
            return [r.to_text() for r in dns.resolver.resolve(domain)]
    except Exception:
        return []
    
def get_ns_info(domain):
    try:
        # Get the A record (IP address) for the NS domain
        a_records = safe_query(domain, 'A')
        if not a_records:
            return f"No A record found for {domain}"
        ip = a_records[0]
        # Use IPWhois to get more information about the IP
        ip_info = IPWhois(ip).lookup_rdap()
        return ip_info
    except Exception as e:
        print(f"Error retrieving NS info for {domain}: {e}")
        return []

print(f"SOA = {safe_query(domain, 'SOA')}")

print("A:", safe_query(domain, 'A'))
print("MX:", safe_query(domain, 'MX'))
print("TXT:", safe_query(domain, 'TXT'))
print("NS:", safe_query(domain, 'NS'))

for ns in safe_query(domain, 'NS'):
    print(f"NS Record: {ns}")
    print("A:", safe_query(ns, 'A'))
    print("NS ip who is Info:", get_ns_info(ns))
    print("NS info", str(safe_query(ns)))
    print()
```
