```python
import os
import gzip
import shutil
import glob

# Caminho da pasta onde os arquivos .gz já estão salvos - após feito o download manualnente
pasta_datasets = r'C:\Users\fsbertoglio\Documents\GitHub\TCC\dns-scripts\caida_data'  # ajuste conforme necessário

# Função para descompactar arquivos .gz
def gunzip_shutil(source_filepath, dest_filepath, block_size=65536):
    with gzip.open(source_filepath, 'rb') as s_file, \
         open(dest_filepath, 'wb') as d_file:
        shutil.copyfileobj(s_file, d_file, block_size)

# Buscar todos os arquivos .gz dentro da pasta especificada
files_gz = glob.glob(os.path.join(pasta_datasets, '*.gz'))

# Descompactar cada arquivo e remover o original .gz
for file in files_gz:
    destino = file[:-3]  # remove ".gz"
    print(f'\nDescompactando: {file}')
    gunzip_shutil(file, destino)
    os.remove(file) 
    print(f'Arquivo descompactado: {destino}')
```
