```python
EU = [
    "Germany", "France", "Netherlands", "Italy", "Spain", "Poland", "Ireland",
    "Luxembourg", "Belgium", "Denmark", "Sweden", "Portugal", "Austria",
    "Hungary", "Czech Republic", "Finland", "Greece", "Cyprus",
    "Estonia", "Lithuania", "Malta", "Romania"
]

# https://european-union.europa.eu/

BRICS = [
    "Brazil", "Russia", "India", "China", "South Africa", "Saudi Arabia"
]
# https://brics2024.ru/en/about-brics

G20 = [
    "United States", "Saudi Arabia", "China", "Germany", "Japan", "United Kingdom",
    "Brazil", "France", "India", "Italy", "Canada", "Australia", "Mexico",
    "South Korea", "Turkey", "Indonesia"
]

# https://www.g20.org/


GICS_SECTORS = [
    'Energy',
    'Materials',
    'Industrials',
    'Consumer Discretionary',
    'Consumer Staples',
    'Health Care',
    'Financials',
    'Information Technology',
    'Communication Services',
    'Utilities',
    'Real Estate'
]
```

```python
import pandas as pd

dff = pd.read_csv(r'.\final_dns_pipeline_multi_ns_cleaned_with_revenue_sorted.csv')


#filtrar out rows where 'revenue' is greater than zero
dff = dff[dff['revenue'] > 0]

#count unique domains
unique_domains = dff['domain'].nunique()

print(f'Unique domains: {unique_domains}')
```

```python
import pandas as pd

# Carregar o dataset
df = pd.read_csv(r'./org_info_pipeline_multi_ns_enriched.csv')





# 1️⃣ Calcular qual o AS mais frequente por domínio
# (em caso de empate, o pandas mantém o primeiro que aparece no dataset original)
dominant_as = (
    df.groupby(['domain', 'as_number'])
      .size()
      .reset_index(name='as_count')
      .sort_values(['domain', 'as_count'], ascending=[True, False])
      .drop_duplicates(subset=['domain'], keep='first')  # mantém o mais frequente
)

# 2️⃣ Juntar de volta com o dataset original para recuperar os outros campos correspondentes à linha dominante
df_dominant = pd.merge(
    dominant_as[['domain', 'as_number']],  # chave do AS dominante por domínio
    df,
    on=['domain', 'as_number'],
    how='left'
).drop_duplicates(subset=['domain'], keep='first')  # garante uma linha por domínio

# 3️⃣ Verificar resultado
print(df_dominant.shape)

#ordem por revenue desc

#only keep rows where revenue is greater than zero
df_dominant = df_dominant[df_dominant['revenue'] > 0]
df_dominant = df_dominant.sort_values(by='revenue', ascending=False)

df_dominant = df_dominant.reset_index(drop=True)
df_dominant.to_csv(r'./unique_org_info_pipeline_multi_ns_dominant_as_enriched.csv', index=False)

display(df_dominant.head(10))
```

```python
# Criar datasets para cada um dos blocos: EU, BRICS, G20 usando como distribuidor o country_name

df_global = df_dominant.copy()
df_eu = df_global[df_global['country_name'].isin(EU)]
df_brics = df_global[df_global['country_name'].isin(BRICS)]
df_g20 = df_global[df_global['country_name'].isin(G20)]

display(df_global)
```

```python
# 1) contabilizar top org_id por soma de revenue em df_global e inclua coluna de contagem, coloque coluna com org_name
def top_orgs_by_revenue(df):
    top_orgs = (
        df.groupby('org_id')
        .agg({'revenue': 'sum', 'domain': 'count', 'org_name': 'first', 'country': 'first'})
        .reset_index()
        .rename(columns={'domain': 'domain_count'})
        .sort_values(by='revenue', ascending=False)
        .reset_index(drop=True)
    )
    return top_orgs

df_global_top_orgs = top_orgs_by_revenue(df_global)
df_eu_top_orgs = top_orgs_by_revenue(df_eu)
df_brics_top_orgs = top_orgs_by_revenue(df_brics)
df_g20_top_orgs = top_orgs_by_revenue(df_g20)

display(df_global_top_orgs)
display(df_eu_top_orgs)
display(df_brics_top_orgs)
display(df_g20_top_orgs)
```
