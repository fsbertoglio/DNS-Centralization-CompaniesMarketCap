```python
import yfinance as yf

tickers = ["AAPL", "MSFT", "PETR4.SA", "VALE3.SA", "AET.L"]

for ticker in tickers:
    info = yf.Ticker(ticker).info
    print(ticker, "->", info.get("website"))
    print(ticker, "->", info.get("longBusinessSummary"))
    print(ticker, "->", info.get("sector"))
    print(ticker, "->", info.get("industry"))
    print(ticker, "->", info.get("country"))
    print(ticker, "->", info.get("city"))
    
```

```python
import glob, os
import pandas as pd

OUTPUT_DIR = "../data/interim/"
FILE_PREFIX = "01_data_after_first_financialdata_"

files = sorted(glob.glob(os.path.join(OUTPUT_DIR, f"{FILE_PREFIX}*.csv")))
data = files[-1]  # Get the most recent file
INPUT_FILE = data  # usado para semear o arquivo de saida caso ele ainda nao exista

print(f"Most recent file: {data}")

df = pd.read_csv(data)

display(df.head(10))

test_df = df[df["website"].isna()]

test_df[["company_name", "stock_symbol", "website"]].head(10)


for company in test_df.head(10).itertuples():
    ticker = company.stock_symbol
    info = yf.Ticker(ticker).info
    print(ticker, "->", info.get("website"))
    print(ticker, "->", info.get("longBusinessSummary"))
    print(ticker, "->", info.get("sector"))
    print(ticker, "->", info.get("industry"))
    print(ticker, "->", info.get("country"))
    print(ticker, "->", info.get("city"))
```
