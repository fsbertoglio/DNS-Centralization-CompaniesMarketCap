# Busca de websites de empresas (com ticker_symbol)

Exemplo de uso direto em notebook das funções de `company_website_finder.py`.


````
import os
# Configure sua API key (ou defina ANTHROPIC_API_KEY no ambiente antes de abrir o notebook)
os.environ["ANTHROPIC_API_KEY"] = "sk-ant-sua-chave-aqui"

from company_website_finder import (
    load_companies_from_csv,
    companies_from_dataframe,
    find_website,
    find_websites_sync,
    find_websites_batch,
    save_results_to_csv,
    print_summary,
)
````

## 1. Carregar empresas do CSV (agora com `ticker_symbol`)

````
companies = load_companies_from_csv("companies_example.csv")
companies[:3]
````

## 2. Testar com UMA empresa (rápido, direto na célula)

````
result = find_website({"name": "Embraer", "ticker_symbol": "EMBR3", "country": "Brasil"})
result
````

## 3. Processar uma lista pequena/média de forma síncrona (resultado na hora)

````
results = find_websites_sync(companies)  # ok para até ~200 empresas
import pandas as pd
df = pd.DataFrame(results)
df
````

## 4. Para listas grandes (500-5000+): usar Batch API (50% de desconto)

Isso bloqueia a célula até o batch terminar (pode levar de minutos até ~1h).

````
results_batch = find_websites_batch(companies)
df_batch = pd.DataFrame(results_batch)
df_batch
````

## 5. Salvar resultado final

````
save_results_to_csv(results_batch, "websites_updated.csv")
print_summary(results_batch)
````

## 6. Se sua lista já está num DataFrame (não em CSV)

Você pode pular o CSV e ir direto do pandas:
```python
import pandas as pd
df_empresas = pd.DataFrame({
    "company_name": ["Embraer", "Vale S.A."],
    "ticker_symbol": ["EMBR3", "VALE3"],
    "country": ["Brasil", "Brasil"],
})
companies = companies_from_dataframe(df_empresas)
results = find_websites_batch(companies)
```
